# Problem 2 - Data Imputations:

```python
df_m = pd.read_csv("/content/drive/MyDrive/Concepts and Technologies of AI/medical_students_dataset.csv")
```

```python
df_m.info()
```

```python
df_m.isnull().sum()
```

```python
numeric_cols = df_m.select_dtypes(include=['float64']).columns
numeric_cols
```

```python
categorical_cols = df_m.select_dtypes(include=['object']).columns
categorical_cols
```

```python
df_m[numeric_cols] = df_m[numeric_cols].fillna(df_m[numeric_cols].median())
# The replaced value does not get affected by extreme values by using median value..

for col in categorical_cols:
    df_m[col] = df_m[col].fillna(df_m[col].mode()[0])
# The missing values are replaced by most frequent occurence of each type...
```

```python
df_m.isnull().sum()
```

```python
df_m.duplicated().sum()
```

```python
df_m = df_m.drop_duplicates()
df_m.duplicated().sum()
```

# Problem - 3:

```python
one_hot_encoding = pd.get_dummies(df_t['Embarked'], prefix='Embarked')
df_t = pd.concat([df_t, one_hot_encoding], axis=1)
df_t = df_t.drop(columns=['Embarked'])
```

```python
print(df_t.head())
```

# Problem - 4:

```python
mean_survival_by_sex = df_t.groupby('Sex')['Survived'].mean()
print(mean_survival_by_sex)
```

```python
import matplotlib.pyplot as plt

mean_survival_by_sex.plot(kind='bar', color=['pink', 'lightblue'])
plt.title('Mean Survival Rate by Sex')
plt.ylabel('Survival Rate')
plt.xlabel('Sex')
plt.show()
```

# Problem - 5:

```python
n_df_t = pd.read_csv("/content/drive/MyDrive/Concepts and Technologies of AI/Titanic-Dataset.csv")
```

```python
mean_survival_by_sex_embarked = n_df_t.groupby(['Sex', 'Embarked'])['Survived'].mean().unstack()
print(mean_survival_by_sex_embarked)
```

```python
import matplotlib.pyplot as plt

mean_survival_by_sex_embarked.plot(kind='bar', color=['lightgreen', 'orange', 'skyblue'])
plt.title('Mean Survival Rate by Sex and Embarked Port')
plt.ylabel('Survival Rate')
plt.xlabel('Sex')
plt.legend(title='Embarked')
plt.show()
```
